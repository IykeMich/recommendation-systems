# 05 — Explain predictions
Reasons come from training-data evidence (bucket fraud rate ≥ 1.5× overall, from at least 50 training transactions). Novelty facts are shown as context, not reasons, because they carry no signal here.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve().parent
sys.path.insert(0, str(ROOT))

import pandas as pd
from src import config
```

```python
from src.explain import explain_transaction
from src.predict import FraudScorer

scorer = FraudScorer.load()
reference = scorer.artifact["explanation_reference"]
print("overall training fraud rate:", reference["overall_fraud_rate"])
test_rows = pd.read_csv(config.PROCESSED_DIR / "test.csv")
for row in test_rows.sort_values("is_fraud", ascending=False).head(3).to_dict("records"):
    risk, features = scorer.score(row)
    explanation = explain_transaction({**row, **features}, reference)
    print(row["transaction_id"], round(risk, 3), [reason["text"] for reason in explanation["reasons"]], explanation["context"])
```
